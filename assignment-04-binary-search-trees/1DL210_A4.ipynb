```python
import unittest

l = (2, (1, None, None), (3, None, None))
r2 = (6, (5, None, None), (7, None, None))

class TestBST(unittest.TestCase):
  def setUp(self):
    self.root = None

  def test_insert(self):
    self.root = insert(self.root, 4)
    self.assertEqual(self.root, (4, None, None))
    self.root = insert(self.root, 2)
    self.assertEqual(self.root, (4, (2, None, None), None))
    self.root = insert(self.root, 3)
    self.assertEqual(self.root, (4, (2, None, (3, None, None)), None))
    self.root = insert(self.root, 1)
    self.assertEqual(self.root, (4, l, None))
    self.root = insert(self.root, 8)
    self.assertEqual(self.root, (4, l, (8, None, None)))
    self.root = insert(self.root, 6)
    self.assertEqual(self.root, (4, l, (8, (6, None, None), None)))
    self.root = insert(self.root, 7)
    self.assertEqual(self.root, (4, l, (8, (6, None, (7, None, None)), None)))
    self.root = insert(self.root, 5)
    self.assertEqual(self.root, (4, l, (8, r2, None)))
    self.root = insert(self.root, 9)
    self.assertEqual(self.root, (4, l, (8, r2, (9, None, None))))

  def test_minimum(self):
    self.assertEqual(minimum(None), None)
    self.assertEqual(minimum(l), 1)
    self.assertEqual(minimum(r2), 5)
    self.test_insert()
    self.assertEqual(minimum(self.root), 1)

  def test_includes(self):
    self.assertEqual(includes(None, 0), False)
    self.assertEqual(includes(l, 0), False)
    self.assertEqual(includes(l, 1), True)
    self.assertEqual(includes(l, 2), True)
    self.assertEqual(includes(l, 3), True)
    self.assertEqual(includes(l, 4), False)

  def test_delete(self):
    self.test_insert()
    self.assertEqual(delete(self.root, 0), self.root)
    self.root = delete(self.root, 4)
    self.assertEqual(self.root, (5, l, (8, (6, None, (7, None, None)), (9, None, None))))
    self.root = delete(self.root, 5)
    self.assertEqual(self.root, (6, l, (8, (7, None, None), (9, None, None))))
    self.root = delete(self.root, 6)
    self.assertEqual(self.root, (7, l, (8, None, (9, None, None))))
    self.root = delete(self.root, 7)
    self.assertEqual(self.root, (8, l, (9, None, None)))
    self.root = delete(self.root, 8)
    self.assertEqual(self.root, (9, l, None))
    self.root = delete(self.root, 9)
    self.assertEqual(self.root, l)

  def test_in_order_walk(self):
    self.assertEqual(in_order_walk(self.root), [])
    self.test_insert()
    self.assertEqual(in_order_walk(self.root), list(range(1, 10)))
    self.setUp()
    self.test_delete()
    self.assertEqual(in_order_walk(self.root), list(range(1, 4)))
```

### 1. BST Insert (2 points)

`insert(t, k)` recursively inserts a key and returns the tree; duplicates are ignored.

```python
def insert(t, k):
  # Base case if our tree its empty
  if t == None:
    return (k,None,None)
  # this means that our key is smaller than the key of the Tree:
  # I dont like that we index it like this i prefere t.key or t.left
  if k < t[0]:
    newLeft = insert(t[1],k)
    return (t[0], newLeft, t[2])
  # here is the other case where the key is greater than the k, that we are comparing with
  elif k > t[0]:
    newRight = insert(t[2],k)
    return (t[0],t[1],newRight)
  # if the key is already on the tree
  else:
    return t

# run the test
unittest.main(argv=[''], defaultTest='TestBST.test_insert', verbosity=2, exit=False)
```

### 2. BST Minimum (1 point)

`minimum(t)` returns the smallest key in the tree, or `None` for an empty tree.

```python
def minimum(t):
  # empty tree
  if t == None:
    return None
  # we search the left tree, as based on the theory of BST is where the smallest value is
  if t[1] == None:
    return t[0]
  # we do this recursively
  return minimum(t[1])

# run the test (this requires that the insert function is implemented)
unittest.main(argv=[''], defaultTest='TestBST.test_minimum', verbosity=2, exit=False)
```

### 3. BST Includes (2 points)

`includes(t, k)` searches the tree and returns a boolean.

```python
def includes(t, k):
  if t == None:
    return False
  # we check if our objective k , if is the same as our root we found it
  if k == t[0]:
    return True
  # Then we compare if its bigger or smaller in order to find if the key is already on the tree
  if k < t[0]:
    return includes(t[1],k)
  else:
    return includes(t[2],k)

# run the test (this requires that the insert function is implemented)
unittest.main(argv=[''], defaultTest='TestBST.test_includes', verbosity=2, exit=False)
```

### 4. BST Delete (4 points)

`delete(t, k)` removes a key if present. For nodes with two children it uses the smallest key in the right subtree as the successor.

```python
def delete(t, k):

  # This is the part where we try to search through items to see where our value is positioned
  if t == None:
    return None

  if k < t[0]:
    newLeft = delete(t[1],k)
    return(t[0],newLeft,t[2])

  if k > t[0]:
    newRight = delete(t[2],k)
    return(t[0],t[1],newRight)

  if t[1] == None:
    return t[2]

  if t[2] == None:
    return t[1]

  # this is key when deleting, as weve seen if a parent have two children its needed
  # to find the successor to take its position otherwise the tree would become invalanced.
  successor = minimum(t[2])
  newRight = delete(t[2], successor)
  return (successor,t[1],newRight)

# run the test (this requires that the insert function is implemented)
unittest.main(argv=[''], defaultTest='TestBST.test_delete', verbosity=2, exit=False)
```

### 5. BST In-order Walk (1 point)

`in_order_walk(t)` returns keys in ascending order.

```python
def in_order_walk(t):
  if t == None:
    return []

  # each of them separates each possibility in its own space l first, then current is the middle one, and then right is the right one
  # that way we can get all the values in order from smallest to bigger
  leftValues = in_order_walk(t[1])

  currentValue = []
  currentValue.append(t[0])

  rightValues = in_order_walk(t[2])

  #we concatenate them to obtain the list in order
  return leftValues + currentValue + rightValues


# run the test (this requires that the insert and delete functions are implemented)
unittest.main(argv=[''], defaultTest='TestBST.test_in_order_walk', verbosity=2, exit=False)
```
